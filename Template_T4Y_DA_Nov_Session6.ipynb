**Conditionals and Iterations**

A conditional statement is one where an action is performed if a prespecified condition is satisfied and another or no action is performed if the condition is not satisfied.

Conditional statements are an easy way to take different actions based on some condition being fulfilled or not.

In Python, we use the *if* statement for the conditional statement.

An iteration is a set of statement that is repeated multiple times.

Iteration will repeat a set of code and perform some action multiple times.

In Python, we use *for* statement for the iteration instruction.


**Advantages:**

* Simplifies coding.

* Readability.

* Efficiency in memory utilization.

## Conditional Statements

**if statement**

```python
#Toss a coin.
coin=np.random.choice(['head','tail'])
print(coin)
```

```python
#what is the value of coin
coin
```

```python
#If the coin lands on heads, Richard wins.
if coin=='head':
  print('Richard wins')
```

**if-else statement**

```python
#If the coin lands on heads, Richard wins. If the coin lands on tail, Emmanuel wins.
if coin =='tail':
  print('Emmanuel wins')
else:
  print('Richard wins')
```

```python
#let's put the toss and the conditional statement together
if np.random.choice(['head','tail'])=='head':
  print('Richard wins')
else:
  print('Emmanuel wins')
```

**if-elif-else statement**

```python
#generate a random number between 1 and 6 (inclusive). It's like rolling a dice.
dice_throw =np.random.randint(1,7)
print(dice_throw)
```

```python
#let's create a conditional statement where if the random number is less than or equal to 3, Richard wins. If the number is equal to 4 or 5, Geraldine wins.
if dice_throw<=3:
  print('Richard wins')
elif dice_throw==4 or dice_throw==5:
  print('Geraldine wins')
else:
  print('Emmanuel wins')
```

```python
#Let's put it all together
dice_throw =np.random.randint(1,7)
if dice_throw<=3:
  print('Richard wins')
elif dice_throw==4 or dice_throw==5:
  print('Geraldine wins')
else:
  print('Emmanuel wins')
```

**Nested if statements**

```python
#Roll a dice. If the number is less than 3, Richard wins.
#If greater than 3, then a coin is tossed and if the coin is heads, then Geraldine wins, else Emmanuel wins.
dice_throw =np.random.randint(1,7)
if dice_throw< 3:
  print('Richard wins')
elif np.random.choice(['head','tail'])=='head':
  print('Geraldine wins')
else:
  print('Emmanuel wins')
```

**While loop basics**

```python
#i = 1. Run a loop such that while i is less than 5, print i
i=1
while i<5:
  print(i)
  i=i+1
```

```python
# Let's look at someone to decide whether they can vote or not. Voting age is 18.
age=1
while age <21:
  if age>18:
    print('You can vote')
  else:
    print ('You cannot vote')
  age=age+1
```

**break statement**


```python
#Let's see how we can exit a while loop using a 'break'
i=1
while i < 10:
  if i ==5:
    print('Exiting the loop at i=',i)
    break
  print(i)
  i=i+1
```
