**Sorting**

```python
# sort the table according to the values in a specific column 'Total'
olympics.sort_values(by=['Total'])
```

```python

# sort the table according to the values in a specific column in a descending order
olympics.sort_values(by=['Total'],ascending=False)
```

You can concatenate commands

```python
# select a two columns and sort accoring to the values of the one
#select the 'Country' and 'Total' columns,then sort by 'Total' in descending order
olympics[['Country','Total']].sort_values(by=['Total'],ascending=False)
```

**Filtering**

```python
# filter rows\instances according to the values of a specific column
# filter the olympics based on 'Gold'
olympics.where(olympics['Gold']>10)
```

```python
olympics.where(olympics['Gold']>10).dropna()
```

```python
# filter rows\instances to retain information with a specific column value between two values
# filter the olympics table based on 'Gold'to show values betwen 2 and 10

olympics.where(olympics['Gold'].between(2,10)).dropna()
```

```python
# filter rows\instances to retain information with a specific column containing a specific string
# filter olympics table where 'Country' contains 'China'

olympics.where(olympics['Country'].str.contains('China')).dropna()
```
